# SVD Geometry of Unit Circle

Chapter 8: Singular Value Decomposition (SVD) · *Mathematics for AI and Machine Learning* by Fuheng Wu

Decompose a 2x2 map into rotation-stretch-rotation and track image axes.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-8/svd-geometry/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import math

import torch

torch.set_printoptions(precision=4)

# The demo's default matrix and probe angle (35 degrees)
A = torch.tensor([[1.8, 0.4],
                  [-0.7, 1.3]], dtype=torch.float64)

# ── A = U Sigma V^T: rotate (V^T), stretch (Sigma), rotate (U) ──────────────
U, S, Vh = torch.linalg.svd(A)
print("singular values =", S.tolist())
print("V (input axes)  =\n", Vh.T)
print("U (output axes) =\n", U)
print("A == U Sigma V^T:", torch.allclose(U @ torch.diag(S) @ Vh, A))

# Each right singular vector maps to a scaled left singular vector
for i in range(2):
    print(f"A v{i+1} = {(A @ Vh[i]).tolist()}  = s{i+1} u{i+1} = {(S[i] * U[:, i]).tolist()}")

# ── Follow the probe x through the three steps ──────────────────────────────
t = math.radians(35)
x = torch.tensor([math.cos(t), math.sin(t)], dtype=torch.float64)
step1 = Vh @ x                 # coordinates in the V basis (still on the unit circle)
step2 = S * step1              # stretched onto the axis-aligned ellipse
step3 = U @ step2              # rotated into place
print("\nprobe x =", x.tolist())
print("V^T x =", step1.tolist(), "| Sigma V^T x =", step2.tolist())
print("U Sigma V^T x =", step3.tolist(), "== A x =", (A @ x).tolist())
print("s2 <= ||Ax|| <= s1:", S[1].item(), "<=", (A @ x).norm().item(), "<=", S[0].item())
```
