# Quadratic Form Geometry

Chapter 7: Symmetric Matrix · *Mathematics for AI and Machine Learning* by Fuheng Wu

Visualize x^T A x level behavior for 2x2 symmetric matrices.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-7/quadratic-form/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default symmetric matrix
A = torch.tensor([[1.8, 0.7],
                  [0.7, 1.2]], dtype=torch.float64)


def q(x):
    """Quadratic form x^T A x = a11 x^2 + 2 a12 x y + a22 y^2."""
    return x @ A @ x


x = torch.tensor([1.0, -0.5], dtype=torch.float64)
print("q([1, -0.5]) =", q(x).item(), "  expanded:",
      (1.8 * 1.0 ** 2 + 2 * 0.7 * 1.0 * -0.5 + 1.2 * 0.25))

# ── The eigen-decomposition rotates q into a sum of squares ─────────────────
evals, Q = torch.linalg.eigh(A)                    # ascending eigenvalues
print("\neigenvalues =", evals.tolist())
print("det(A) =", torch.linalg.det(A).item(), " trace(A) =", A.trace().item())
y = Q.T @ x                                       # coordinates along the eigenvectors
print("q(x) = sum lambda_i y_i^2 =", (evals * y ** 2).sum().item())

# Both eigenvalues > 0, so every level set q(x) = c is an ellipse.
# Its semi-axes along the eigenvectors have length sqrt(c / lambda_i).
c = 1.0
print("\nlevel set q = 1: semi-axes", (c / evals).sqrt().tolist())

# ── The gradient is 2 A x, always perpendicular to the level curve ──────────
x = x.clone().requires_grad_(True)
q(x).backward()
print("autograd grad =", x.grad.tolist(), " 2 A x =", (2 * A @ x.detach()).tolist())
```
