Inspect orthonormal eigenvectors and diagonalization of symmetric matrices.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-7/spectral-theorem/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default symmetric matrix
A = torch.tensor([[2.0, 0.6],
                  [0.6, 1.0]], dtype=torch.float64)

# ── Spectral theorem: A = Q Lambda Q^T with orthonormal Q ───────────────────
evals, Q = torch.linalg.eigh(A)          # eigh: for symmetric matrices, real and ordered
L = torch.diag(evals)
print("eigenvalues =", evals.tolist())
print("Q =\n", Q)
print("Q^T Q == I       :", torch.allclose(Q.T @ Q, torch.eye(2, dtype=A.dtype)))
print("A == Q Lambda Q^T:", torch.allclose(Q @ L @ Q.T, A))
for lam, qv in zip(evals, Q.T):
    print(f"||A q - {lam.item():.4f} q|| =", (A @ qv - lam * qv).norm().item())

# ── As a sum of rank-one projections: A = sum lambda_i q_i q_i^T ────────────
parts = [lam * torch.outer(qv, qv) for lam, qv in zip(evals, Q.T)]
print("\nA == sum lambda_i q_i q_i^T:", torch.allclose(sum(parts), A))

# ── A non-symmetric matrix has no such orthonormal eigenbasis ───────────────
B = torch.tensor([[2.0, 1.5],
                  [0.0, 1.0]], dtype=torch.float64)
_, V = torch.linalg.eig(B)
V = V.real
print("\nnon-symmetric B: angle between its eigenvectors (deg) =",
      torch.rad2deg(torch.arccos((V[:, 0] @ V[:, 1]).abs() / (V[:, 0].norm() * V[:, 1].norm()))).item())
```
