```python
import torch

u = torch.tensor([1.0, 2.0, -1.0])
v = torch.tensor([3.0, 0.5, 2.0])

# Outer product: (u v^T)_{ij} = u_i v_j  — a rank-1 matrix
outer = torch.outer(u, v)

print(outer)
print("entry (i,j) = u[i] * v[j]:", outer[0, 1].item(), "==", (u[0] * v[1]).item())
```
