# L1 vs L2 Regularization

Chapter 1: Vector Space and Inner Product · *Mathematics for AI and Machine Learning* by Fuheng Wu

Constraint geometry and lasso/ridge paths—why L1 hits axes.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-1/l1-l2-geometry/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# Unregularized least-squares center θ̂ (same as the demo contour minimum)
theta_hat = torch.tensor([1.08, 0.38])
lam = 0.35

# L1 (Lasso): soft-threshold each coordinate toward 0
# sign(θ̂_i) * max(|θ̂_i| − λ, 0)
lasso = torch.sign(theta_hat) * torch.clamp(theta_hat.abs() - lam, min=0.0)

# L2 (Ridge): shrink toward origin — θ / (1 + λ)
ridge = theta_hat / (1.0 + lam)

print("θ̂ =", theta_hat.tolist())
print("Lasso(λ) =", lasso.tolist())
print("Ridge(λ) =", ridge.tolist())
print("L1 norm Lasso:", lasso.abs().sum().item())
print("L2 norm Ridge:", ridge.norm().item())
```
