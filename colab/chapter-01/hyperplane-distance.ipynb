3D hyperplane θᵀx = b, foot of perpendicular, and signed distance d⊥.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-1/hyperplane-distance/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# Hyperplane θ^T x = b  (θ is the normal vector)
theta = torch.tensor([0.6, -0.8, 0.2])
b = torch.tensor(1.0)
x = torch.tensor([1.0, 2.0, 3.0])

# Signed distance from x to the plane:
#   d = (θ^T x − b) / ||θ||
numer = torch.dot(theta, x) - b
denom = torch.linalg.norm(theta)
signed_dist = numer / denom

print("signed distance:", signed_dist.item())
```
