```python
import torch

# Vectors u, v in R^2 — same idea as the parallelogram in the demo
u = torch.tensor([1.5, 1.0])
v = torch.tensor([2.2, -0.6])

# Component-wise addition: c = u + v (diagonal of the parallelogram)
c = u + v

print("u + v =", c.tolist())
```
