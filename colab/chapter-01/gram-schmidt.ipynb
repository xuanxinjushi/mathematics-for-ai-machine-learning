# Gram-Schmidt Orthonormalization

Chapter 1: Vector Space and Inner Product · *Mathematics for AI and Machine Learning* by Fuheng Wu

Build an orthonormal basis from two input vectors step by step.

```python
import torch


def gram_schmidt(vectors: torch.Tensor) -> torch.Tensor:
    """Orthonormalize rows of `vectors` (m × n)."""
    out = []
    for i in range(vectors.size(0)):
        v = vectors[i].clone()
        # Subtract projections onto already-orthonormalized rows
        for u in out:
            v = v - torch.dot(v, u) * u
        # Normalize to unit length
        out.append(v / torch.linalg.norm(v))
    return torch.stack(out)


V = torch.tensor([[1.0, 1.0], [2.0, 0.5]])
Q = gram_schmidt(V)

print("orthonormal rows:\n", Q)
# For orthonormal rows: Q Q^T ≈ I
print("Q Q^T ≈ I:\n", Q @ Q.T)
```
