```python
import torch

u = torch.tensor([1.5, 1.0])
v = torch.tensor([2.2, -0.6])

# Inner product ⟨u,v⟩ = u^T v = sum_i u_i v_i
dot = torch.inner(u, v)

# Euclidean norms ||u||_2, ||v||_2
norm_u = torch.linalg.norm(u)
norm_v = torch.linalg.norm(v)

# cos θ = ⟨u,v⟩ / (||u|| ||v||)  — Cauchy–Schwarz bounds |cos θ| ≤ 1
cos_theta = dot / (norm_u * norm_v)

# acos needs inputs in [-1, 1]; clamp avoids NaNs from rounding error
angle = torch.acos(cos_theta.clamp(-1.0, 1.0))

print("⟨u,v⟩ =", dot.item())
print("cos θ =", cos_theta.item())
print("θ (rad) =", angle.item())
```
