```python
import torch
import torch.nn as nn

# Simple 1D regression: y ≈ w x + b  (matches the demo's 2D slice)
X = torch.tensor([[1.0], [2.0], [3.0]])   # feature column
y = torch.tensor([[2.0], [4.0], [5.0]])   # targets

model = nn.Linear(1, 1)                   # one input, one output
lr = 0.1                                  # learning rate η
opt = torch.optim.SGD(model.parameters(), lr=lr)
loss_fn = nn.MSELoss()                    # mean squared error

for _ in range(200):
    pred = model(X)                       # forward: ŷ = w x + b
    loss = loss_fn(pred, y)               # (1/n) Σ (y − ŷ)²
    opt.zero_grad()                       # clear old gradients
    loss.backward()                       # ∂loss/∂w, ∂loss/∂b
    opt.step()                            # gradient descent update

w, b = model.weight.item(), model.bias.item()
print(f"y ≈ {w:.3f} x + {b:.3f}")
```
