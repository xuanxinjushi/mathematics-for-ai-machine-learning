```python
import torch

u = torch.tensor([1.5, 1.0])
v = torch.tensor([2.2, -0.6])

dot = torch.dot(u, v)
bound = torch.linalg.norm(u) * torch.linalg.norm(v)

# Cauchy–Schwarz: |⟨u,v⟩| ≤ ||u|| ||v||
print("|⟨u,v⟩| =", dot.abs().item())
print("||u|| ||v|| =", bound.item())
print("Cauchy–Schwarz holds:", dot.abs().item() <= bound.item() + 1e-6)
```
