# Power Iteration Convergence

Chapter 6: Geometric Transformation and Eigen Decomposition · *Mathematics for AI and Machine Learning* by Fuheng Wu

Iterate x_{k+1}=Ax_k/||Ax_k|| and track dominant eigenvector convergence.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-6/power-iteration/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's default matrix, start vector and number of iterations
A = torch.tensor([[2.0, 0.3],
                  [0.4, 1.2]], dtype=torch.float64)
x = torch.tensor([1.0, 0.25], dtype=torch.float64)
n_iter = 12

# The answer we are converging to
evals, evecs = torch.linalg.eig(A)
order = evals.real.abs().argsort(descending=True)
lam1, lam2 = evals.real[order]
v1 = evecs.real[:, order[0]]
v1 = v1 * torch.sign(v1[0])
print(f"dominant eigenvalue = {lam1.item():.6f}, |lambda2/lambda1| = {(lam2 / lam1).abs().item():.4f}")

# ── x_{k+1} = A x_k / ||A x_k||, Rayleigh quotient as the eigenvalue estimate
print("\n  k   Rayleigh x^T A x     angle error to v1")
for k in range(1, n_iter + 1):
    y = A @ x
    x = y / y.norm()
    rayleigh = x @ A @ x
    err = (x * torch.sign(x[0]) - v1).norm()
    print(f"{k:3d}   {rayleigh.item():.10f}     {err.item():.3e}")

# The error shrinks by about |lambda2/lambda1| per step
print("\nexpected shrink factor per step:", (lam2 / lam1).abs().item())
```
