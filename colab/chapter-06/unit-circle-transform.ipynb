Map the unit circle by a 2x2 matrix and inspect stretching and rotation.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-6/unit-circle-transform/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import math

import torch

torch.set_printoptions(precision=4)

# The demo's default matrix; its columns are where e1 and e2 land
A = torch.tensor([[1.8, 0.6],
                  [-0.4, 1.2]], dtype=torch.float64)
print("A e1 =", A[:, 0].tolist(), "  A e2 =", A[:, 1].tolist())

# ── Push the unit circle through A ──────────────────────────────────────────
t = torch.linspace(0, 2 * math.pi, 721, dtype=torch.float64)
circle = torch.stack([t.cos(), t.sin()])          # 2 x N points with ||x|| = 1
ellipse = A @ circle
lengths = ellipse.norm(dim=0)

# The longest and shortest images are the ellipse's semi-axes = singular values
print("longest  ||Ax|| on the circle =", lengths.max().item())
print("shortest ||Ax|| on the circle =", lengths.min().item())
print("singular values of A          =", torch.linalg.svdvals(A).tolist())

# ── Area: the unit square (area 1) maps to a parallelogram of area |det A| ──
det = torch.linalg.det(A).item()
print("\ndet(A) =", det, "-> areas scale by", abs(det), "| orientation",
      "kept" if det > 0 else "flipped")
# The ellipse's area is pi * s1 * s2 = pi * |det A|
s = torch.linalg.svdvals(A)
print("ellipse area pi*s1*s2 =", (math.pi * s.prod()).item(), " pi*|det A| =", math.pi * abs(det))
```
