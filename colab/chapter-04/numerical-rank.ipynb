# Numerical Rank via Singular Values

Chapter 4: QR Decomposition and Numerical Rank · *Mathematics for AI and Machine Learning* by Fuheng Wu

Explore threshold-dependent effective rank from singular values.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-4/numerical-rank/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.manual_seed(0)

# The demo's default sliders: singular values s1, s2 and the threshold tau
s1, s2, tau = 3.0, 0.15, 0.2

# ── Build A = U diag(s) V^T with random orthogonal U, V ─────────────────────
U, _ = torch.linalg.qr(torch.randn(2, 2, dtype=torch.float64))
V, _ = torch.linalg.qr(torch.randn(2, 2, dtype=torch.float64))
A = U @ torch.diag(torch.tensor([s1, s2], dtype=torch.float64)) @ V.T
print("A =\n", A)

# The entries of A hide the near-dependence; the SVD exposes it
S = torch.linalg.svdvals(A)
print("singular values =", S.tolist())

# ── Numerical rank = number of singular values above tau ────────────────────
rank_tau = int((S > tau).sum())
print(f"rank with tau = {tau}:", rank_tau)
print("rank with the default tolerance:", torch.linalg.matrix_rank(A).item())
print("rank via torch (atol = tau):", torch.linalg.matrix_rank(A, atol=tau).item())
print("condition number s1/s2 =", (S[0] / S[1]).item())

# ── Dropping s2 gives the best rank-1 approximation (Eckart-Young) ──────────
Uf, Sf, Vhf = torch.linalg.svd(A)
A1 = Sf[0] * torch.outer(Uf[:, 0], Vhf[0])
print("\n||A - A_1||_2 =", torch.linalg.matrix_norm(A - A1, ord=2).item(), "(equals s2)")
```
