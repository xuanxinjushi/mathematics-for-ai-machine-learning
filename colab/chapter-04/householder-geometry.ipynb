# Householder Reflection Geometry

Chapter 4: QR Decomposition and Numerical Rank · *Mathematics for AI and Machine Learning* by Fuheng Wu

Reflect vectors across a hyperplane and observe norm preservation.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-4/householder-geometry/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default sliders: mirror normal u and the vector v to reflect
u = torch.tensor([1.2, 2.1], dtype=torch.float64)
v = torch.tensor([2.4, 0.5], dtype=torch.float64)

# ── Householder matrix  H = I - 2 u u^T / (u^T u) ───────────────────────────
H = torch.eye(2, dtype=u.dtype) - 2 * torch.outer(u, u) / (u @ u)
Hv = H @ v
print("H =\n", H)
print("H v =", Hv.tolist())

# A reflection keeps lengths, is its own inverse and flips orientation
print("||v|| =", v.norm().item(), " ||Hv|| =", Hv.norm().item())
print("H^T H == I:", torch.allclose(H.T @ H, torch.eye(2, dtype=u.dtype)),
      "| H H == I:", torch.allclose(H @ H, torch.eye(2, dtype=u.dtype)),
      "| det H =", torch.linalg.det(H).item())

# The midpoint of v and Hv lies on the mirror (orthogonal to u)
print("u · (v + Hv)/2 (should be 0) =", (u @ (v + Hv) / 2).item())

# ── The QR use: pick u so that H sends a vector x onto the e1 axis ──────────
x = torch.tensor([3.0, 4.0], dtype=torch.float64)
e1 = torch.tensor([1.0, 0.0], dtype=torch.float64)
alpha = -torch.sign(x[0]) * x.norm()          # sign choice avoids cancellation
w = x - alpha * e1
Hx = (torch.eye(2, dtype=x.dtype) - 2 * torch.outer(w, w) / (w @ w)) @ x
print("\nH x =", Hx.tolist(), "-> all of ||x|| =", x.norm().item(), "lands on the first axis")
```
