Relate eigenvalues to triangular Schur form entries for 2x2 examples.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-9/schur-form/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import numpy as np
import scipy.linalg
import torch

np.set_printoptions(precision=4, suppress=True)

# The demo's default matrix
A = np.array([[1.4, 0.9],
              [-0.7, 1.1]])
tr, det = np.trace(A), np.linalg.det(A)
print("trace =", tr, " det =", round(det, 6), " discriminant tr^2 - 4 det =", round(tr ** 2 - 4 * det, 6))
print("eigenvalues:", np.linalg.eigvals(A))           # a complex pair

# ── Real Schur form: A = Z T Z^T, T quasi-triangular (2x2 block for a pair) ─
T, Z = scipy.linalg.schur(A, output="real")
print("\nreal Schur T =\n", T)
print("Z orthogonal:", np.allclose(Z.T @ Z, np.eye(2)), "| A == Z T Z^T:", np.allclose(Z @ T @ Z.T, A))

# ── Complex Schur form: upper triangular, eigenvalues on the diagonal ───────
Tc, Zc = scipy.linalg.schur(A, output="complex")
print("\ncomplex Schur T =\n", Tc)
print("diag(T) =", np.diag(Tc), "| A == Z T Z^H:", np.allclose(Zc @ Tc @ Zc.conj().T, A))

# ── With real eigenvalues the real Schur form is already triangular ─────────
B = torch.tensor([[1.4, 0.9],
                  [0.7, 1.1]], dtype=torch.float64)
lam, V = torch.linalg.eig(B)
q1 = V.real[:, 0] / V.real[:, 0].norm()                # one unit eigenvector
Q = torch.stack([q1, torch.stack([-q1[1], q1[0]])], dim=1)
TB = Q.T @ B @ Q
print("\nB with real eigenvalues", lam.real.tolist())
print("Q^T B Q =\n", TB.numpy(), "\n-> lower-left entry", TB[1, 0].item(), "(triangular)")
```
