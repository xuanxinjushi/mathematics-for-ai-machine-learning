# Polar Decomposition Explorer

Chapter 9: Matrix Decompositions and Beyond · *Mathematics for AI and Machine Learning* by Fuheng Wu

Split a matrix into orthogonal and symmetric factors and inspect reconstruction.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-9/polar-decomposition/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default matrix
A = torch.tensor([[1.7, 0.8],
                  [-0.6, 1.2]], dtype=torch.float64)

# ── Polar decomposition A = Q S from the SVD A = U Sigma V^T ────────────────
U, Sig, Vh = torch.linalg.svd(A)
Q = U @ Vh                              # orthogonal part ("rotation")
S = Vh.T @ torch.diag(Sig) @ Vh         # symmetric positive semidefinite part ("stretch")

print("Q =\n", Q)
print("S =\n", S)
print("A == Q S       :", torch.allclose(Q @ S, A))
print("Q^T Q == I     :", torch.allclose(Q.T @ Q, torch.eye(2, dtype=A.dtype)))
print("S symmetric    :", torch.allclose(S, S.T), "| eigenvalues of S =", torch.linalg.eigvalsh(S).tolist())
print("singular values of A =", Sig.tolist(), "(the stretches)")
print("det A =", torch.linalg.det(A).item(), " det Q =", torch.linalg.det(Q).item())

# S is also the matrix square root of A^T A
print("\nS^2 == A^T A:", torch.allclose(S @ S, A.T @ A))

# ── Q is the orthogonal matrix closest to A (Procrustes) ────────────────────
torch.manual_seed(0)
best = torch.linalg.matrix_norm(A - Q).item()
others = []
for _ in range(2000):
    R, _ = torch.linalg.qr(torch.randn(2, 2, dtype=torch.float64))
    others.append(torch.linalg.matrix_norm(A - R).item())
print(f"||A - Q||_F = {best:.6f}; best of 2000 random orthogonal matrices = {min(others):.6f}")
```
