Split a vector into projection and orthogonal residual components.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-3/orthogonal-decomposition/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's default sliders: direction u and the vector v to split
u = torch.tensor([2.2, 1.0], dtype=torch.float64)
v = torch.tensor([-0.5, 2.4], dtype=torch.float64)

# ── Projection onto span{u}:  p = (u·v / u·u) u ─────────────────────────────
p = (torch.dot(u, v) / torch.dot(u, u)) * u
# ── Orthogonal residual:  r = v - p  (lives in u-perp) ──────────────────────
r = v - p

print("projection p =", p.tolist())
print("residual   r =", r.tolist())
print("p + r == v  :", torch.allclose(p + r, v))
print("u · r (should be 0)  =", torch.dot(u, r).item())

# Pythagoras: ||v||^2 = ||p||^2 + ||r||^2 because p is orthogonal to r
print("||v||^2 =", v.dot(v).item(), " ||p||^2 + ||r||^2 =", (p.dot(p) + r.dot(r)).item())

# ── Same thing with the projection matrix P = u u^T / (u^T u) ───────────────
P = torch.outer(u, u) / torch.dot(u, u)
print("\nP =\n", P)
print("P v == p:", torch.allclose(P @ v, p), "| P^2 == P:", torch.allclose(P @ P, P),
      "| P^T == P:", torch.allclose(P.T, P))
```
