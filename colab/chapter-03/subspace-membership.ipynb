# Subspace Membership Checker

Chapter 3: Subspaces and Orthogonality · *Mathematics for AI and Machine Learning* by Fuheng Wu

Check if a target vector lies in the span of two basis vectors in 2D.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-3/subspace-membership/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default sliders: two basis vectors b1, b2 and a target v (all in R^2)
b1 = torch.tensor([1.8, 0.9], dtype=torch.float64)
b2 = torch.tensor([-0.6, 1.6], dtype=torch.float64)
v = torch.tensor([0.8, 2.0], dtype=torch.float64)

# ── Is v in span{b1, b2}?  Solve B c = v in the least-squares sense ──────────
B = torch.stack([b1, b2], dim=1)          # columns are the basis vectors
c = torch.linalg.lstsq(B, v.unsqueeze(1)).solution.squeeze(1)
residual = v - B @ c

print("coefficients c =", c)
print("B c            =", B @ c)
print("residual ||v - Bc|| =", residual.norm().item())
print("rank(B) =", torch.linalg.matrix_rank(B).item())

# Two independent vectors in R^2 span the whole plane, so the residual is ~0.
# ── Now make b2 parallel to b1: the span collapses to a line ────────────────
b2_parallel = 0.5 * b1
B_line = torch.stack([b1, b2_parallel], dim=1)
c_line = torch.linalg.lstsq(B_line, v.unsqueeze(1), driver="gelsd").solution.squeeze(1)
res_line = v - B_line @ c_line

print("\nwith b2 = 0.5 b1: rank(B) =", torch.linalg.matrix_rank(B_line).item())
print("residual ||v - Bc|| =", res_line.norm().item(), "-> v is not on that line")
```
