# Rotary Position Embedding (RoPE)

Chapter 3: Subspaces and Orthogonality · *Mathematics for AI and Machine Learning* by Fuheng Wu

Rotate query/key pairs by position; the attention score depends only on the relative distance n − m.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-3/rope/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.manual_seed(0)
d = 8                                     # head dimension (d/2 rotation pairs)
theta = 10000 ** (-torch.arange(0, d, 2) / d)


def rope(x, pos):
    # Treat each pair (x[2i], x[2i+1]) as a complex number and rotate it by pos * theta_i
    z = torch.view_as_complex(x.reshape(-1, 2).contiguous())
    return torch.view_as_real(z * torch.polar(torch.ones_like(theta), pos * theta)).flatten()


q, k = torch.randn(d), torch.randn(d)

# Same relative distance n - m = 4 at different absolute positions
for m, n in [(0, 4), (10, 14), (100, 104)]:
    print(f"m={m:3d}, n={n:3d}:  q_m . k_n = {rope(q, m) @ rope(k, n):.6f}")

# Rotations preserve length (they are orthonormal)
print("||q|| =", round(q.norm().item(), 6), " ||q_m|| =", round(rope(q, 37).norm().item(), 6))
```
