# Null Space and Column Space View

Chapter 3: Subspaces and Orthogonality · *Mathematics for AI and Machine Learning* by Fuheng Wu

Inspect rank, nullity, and geometric directions for a 2x2 matrix.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-3/null-column-space/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default matrix: the second column is 1.5x the first, so rank 1
A = torch.tensor([[1.0, 1.5],
                  [2.0, 3.0]], dtype=torch.float64)

# ── Rank and nullity from the SVD ───────────────────────────────────────────
U, S, Vh = torch.linalg.svd(A)
tol = S.max() * max(A.shape) * torch.finfo(A.dtype).eps
rank = int((S > tol).sum())
print("singular values =", S)
print("rank =", rank, " nullity =", A.shape[1] - rank, " (rank + nullity = n =", A.shape[1], ")")

# ── Column space: the left singular vectors with nonzero singular values ────
col_basis = U[:, :rank]
print("\ncolumn space basis:\n", col_basis)

# ── Null space: right singular vectors with zero singular values ────────────
null_basis = Vh[rank:].T
print("null space basis:\n", null_basis)
print("A @ null vector (should be 0):", (A @ null_basis).squeeze().tolist())

# Any b outside the column space has no exact solution of A x = b
b = torch.tensor([1.0, 0.0], dtype=torch.float64)
in_col = (b - col_basis @ (col_basis.T @ b)).norm()
print("\ndistance of b =", b.tolist(), "from the column space =", in_col.item())
```
