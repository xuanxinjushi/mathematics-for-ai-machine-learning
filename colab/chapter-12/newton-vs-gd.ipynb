# Newton vs Gradient Descent

Chapter 12: Optimization Methods · *Mathematics for AI and Machine Learning* by Fuheng Wu

Compare first-order and second-order updates on a curved one-dimensional objective.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-12/newton-vs-gd/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's objective and defaults: f(x) = x^4/4 + x^2/2, eta = 0.15, x0 = 2.2, 12 steps
eta, x0, K = 0.15, 2.2, 12


def f(x):
    return 0.25 * x ** 4 + 0.5 * x ** 2


def grad_and_hess(x):
    x = x.detach().requires_grad_(True)
    (g,) = torch.autograd.grad(f(x), x, create_graph=True)
    (h,) = torch.autograd.grad(g, x)
    return g.detach(), h.detach()


# ── Gradient descent x <- x - eta f'  vs  Newton x <- x - f'/f'' ────────────
xg = xn = torch.tensor(x0, dtype=torch.float64)
print(" k      GD x          GD f(x)        Newton x        Newton f(x)")
for k in range(K + 1):
    print(f"{k:2d}  {xg.item():+.6e}  {f(xg).item():.6e}   {xn.item():+.6e}  {f(xn).item():.6e}")
    g, _ = grad_and_hess(xg)
    xg = xg - eta * g
    g, h = grad_and_hess(xn)
    xn = xn - g / h

# Newton's error is squared each step near the minimum (quadratic convergence)
print("\nf'(x) = x^3 + x, f''(x) = 3x^2 + 1 -> Newton step x - (x^3 + x)/(3x^2 + 1) = 2x^3/(3x^2 + 1)")
print("near 0 that is about 2 x^3: the number of correct digits roughly triples per step")
```
