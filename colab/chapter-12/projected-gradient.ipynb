# Projected Gradient on Box Constraint

Chapter 12: Optimization Methods · *Mathematics for AI and Machine Learning* by Fuheng Wu

Show unconstrained steps followed by projection onto a feasible interval.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-12/projected-gradient/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's defaults: minimize 0.5 (x - c)^2 over the box [-1, 1]
c, eta, x0 = 2.0, 0.3, -2.5
lo, hi = -1.0, 1.0


def project(x):
    return x.clamp(lo, hi)            # nearest point of the interval


x_free = torch.tensor(x0, dtype=torch.float64)
x_proj = torch.tensor(x0, dtype=torch.float64)
print(" k   unconstrained   projected")
for k in range(41):
    if k in (0, 1, 2, 3, 5, 10, 20, 40):
        print(f"{k:2d}   {x_free.item():+.6f}      {x_proj.item():+.6f}")
    x_free = x_free - eta * (x_free - c)
    x_proj = project(x_proj - eta * (x_proj - c))   # step, then project

x_star = min(max(c, lo), hi)
print("\nconstrained minimizer clip(c, -1, 1) =", x_star, " unconstrained minimizer c =", c)

# At the solution the gradient is not zero, but it points out of the box:
g = x_star - c
print("gradient at x* =", g, "-> moving inside the box (x < 1) would only increase f")
# Fixed point check: projecting one more step leaves x* unchanged
print("x* == project(x* - eta grad):", project(torch.tensor(x_star - eta * g)).item() == x_star)
```
