# Gradient Descent Step Size Explorer

Chapter 12: Optimization Methods · *Mathematics for AI and Machine Learning* by Fuheng Wu

Inspect convergence or divergence of gradient descent as learning rate changes.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-12/gd-stepsize/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's defaults: f(x) = 0.5 a x^2, step size eta, start x0
a, eta, x0 = 1.5, 0.4, 2.0

# ── x_{k+1} = x_k - eta f'(x_k) = (1 - eta a) x_k ───────────────────────────
m = 1 - eta * a
print(f"per-step factor 1 - eta a = {m}; converges iff |1 - eta a| < 1, i.e. 0 < eta < 2/a = {2 / a:.4f}")

x = torch.tensor(x0, dtype=torch.float64, requires_grad=True)
opt = torch.optim.SGD([x], lr=eta)
for k in range(8):
    print(f"k={k}: x = {x.item():+.6f}  closed form x0 m^k = {x0 * m ** k:+.6f}")
    opt.zero_grad()
    (0.5 * a * x ** 2).backward()
    opt.step()

# ── Sweep the step size ─────────────────────────────────────────────────────
print("\n  eta    1 - eta a    |x_70|        behaviour")
for e in (0.1, 0.4, 2 / a / 2, 1.0, 2 / a, 1.5):
    f = 1 - e * a
    x70 = abs(x0 * f ** 70)
    kind = ("oscillates and diverges" if f < -1 else "stuck (period 2)" if abs(f + 1) < 1e-12
            else "converges, oscillating" if f < 0 else "one step to the minimum" if f == 0 else "converges monotonically")
    print(f"{e:5.3f}   {f:+8.4f}   {x70:11.4e}   {kind}")
```
