Factor a 2x2 matrix into L and U and verify the reconstruction.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-5/lu-steps/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's default 2x2 matrix
A = torch.tensor([[2.0, 1.0],
                  [3.0, 2.5]], dtype=torch.float64)

# ── One elimination step: row2 <- row2 - l21 * row1 ─────────────────────────
l21 = A[1, 0] / A[0, 0]                    # multiplier = 3 / 2
U = A.clone()
U[1] -= l21 * U[0]
L = torch.tensor([[1.0, 0.0],
                  [l21, 1.0]], dtype=torch.float64)

print("multiplier l21 =", l21.item())
print("L =\n", L)
print("U =\n", U)
print("A == L U:", torch.allclose(L @ U, A))

# ── torch.linalg.lu uses partial pivoting: P A = L U, P swaps rows ──────────
P, L2, U2 = torch.linalg.lu(A)
print("\ntorch.linalg.lu (with pivoting):")
print("P =\n", P)
print("L =\n", L2)
print("U =\n", U2)
print("A == P L U:", torch.allclose(P @ L2 @ U2, A))
# |3| > |2| in the first column, so pivoting puts row 2 on top first.
```
