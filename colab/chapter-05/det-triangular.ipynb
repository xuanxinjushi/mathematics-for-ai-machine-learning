# Determinant from Triangular Factors

Chapter 5: Square Matrix and LU Decomposition · *Mathematics for AI and Machine Learning* by Fuheng Wu

Connect det(A) with diagonal products of triangular factors.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-5/det-triangular/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's default 2x2 matrix
A = torch.tensor([[2.5, 0.8],
                  [-1.4, 1.9]], dtype=torch.float64)

# ── det(A) = det(L) det(U) = 1 * prod(diag U) for A = L U ───────────────────
l21 = A[1, 0] / A[0, 0]
U = A.clone()
U[1] -= l21 * U[0]
print("U =", U.tolist())
print("prod(diag U) =", U.diagonal().prod().item())
print("ad - bc      =", (A[0, 0] * A[1, 1] - A[0, 1] * A[1, 0]).item())
print("torch det    =", torch.linalg.det(A).item())

# ── With pivoting, each row swap flips the sign: det A = det(P) prod(diag U) ─
P, L, U2 = torch.linalg.lu(A)
sign = torch.linalg.det(P)                   # +1 or -1
print("\npivoted: det(P) =", sign.item(), " prod(diag U) =", U2.diagonal().prod().item(),
      " product =", (sign * U2.diagonal().prod()).item())

# ── For big matrices use log|det| to avoid overflow/underflow ───────────────
torch.manual_seed(0)
B = torch.randn(400, 400, dtype=torch.float64)
sign_b, logabs = torch.linalg.slogdet(B)
_, _, Ub = torch.linalg.lu(B)
print("\n400x400: sum log|U_ii| =", Ub.diagonal().abs().log().sum().item(), " slogdet =", logabs.item())
```
