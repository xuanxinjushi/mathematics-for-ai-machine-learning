# 2D Langevin Dynamics

Chapter 19: Stochastic Differential Equations · *Mathematics for AI and Machine Learning* by Fuheng Wu

A live 3D demo: particles on a ring relax under dX = -X dt + sqrt(2) dW to the standard Gaussian, with E|X|^2 following 2 + 7 e^{-2t}.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-19/langevin-2d/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import math

import torch

torch.manual_seed(42)

# The demo's defaults: dX = -X dt + sqrt(2) dW in 2D (overdamped Langevin for N(0, I)), T = 6, dt = 0.02
T = 6.0
dt = 0.02
n_demo = 15
steps = round(T / dt)

# Particles start far from equilibrium, on a ring of radius 3
n = 20000
ang = 2 * math.pi * torch.rand(n, dtype=torch.float64)
x = 3 * torch.stack([ang.cos(), ang.sin()], 1)

print("   t     mean |x|^2   theory 2 + (9 - 2) e^{-2t}")
for k in range(steps + 1):
    t = k * dt
    if k in (0, 25, 50, 100, 200, steps):
        print(f"{t:5.2f}    {(x ** 2).sum(1).mean().item():.4f}       {2 + 7 * math.exp(-2 * t):.4f}")
    x = x + (-x) * dt + math.sqrt(2 * dt) * torch.randn_like(x)

print("\nfinal covariance =", [[round(v, 3) for v in r] for r in torch.cov(x.T).tolist()], "(target I)")
# The demo draws only 15 particles, so its cloud is a rough version of this.
print(f"with {n_demo} particles the sample variance is noisy: e.g.", [round(v, 2) for v in x[:n_demo].var(0).tolist()])
```
