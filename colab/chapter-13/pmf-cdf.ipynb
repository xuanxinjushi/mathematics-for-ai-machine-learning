# Binomial PMF & CDF

Chapter 13: Probability and Random Variables · *Mathematics for AI and Machine Learning* by Fuheng Wu

Compare binomial probability mass with the cumulative step function.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-13/pmf-cdf/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import math

import torch
from torch.distributions import Binomial

# The demo's defaults: Binomial(n = 12, p = 0.35)
n, p = 12, 0.35
dist = Binomial(n, probs=torch.tensor(p, dtype=torch.float64))
k = torch.arange(n + 1, dtype=torch.float64)

pmf = dist.log_prob(k).exp()
cdf = pmf.cumsum(0)
by_hand = torch.tensor([math.comb(n, int(i)) * p ** i * (1 - p) ** (n - i) for i in k])

print(" k    P(X = k)    F(k) = P(X <= k)")
for i in range(n + 1):
    print(f"{i:2d}    {pmf[i].item():.5f}     {cdf[i].item():.5f}")
print("\nPMF matches C(n,k) p^k (1-p)^(n-k):", torch.allclose(pmf, by_hand))
print("sum of PMF = F(n) =", cdf[-1].item())

# ── Mean and variance from the PMF vs np and np(1 - p) ──────────────────────
mean = (k * pmf).sum().item()
var = ((k - mean) ** 2 * pmf).sum().item()
print(f"E[X] = {mean:.4f} (np = {n * p})   Var(X) = {var:.4f} (np(1-p) = {n * p * (1 - p):.4f})")

# Interval probabilities come from CDF differences
print("P(3 <= X <= 6) = F(6) - F(2) =", (cdf[6] - cdf[2]).item())
# The median is the smallest k with F(k) >= 1/2
print("median =", int((cdf >= 0.5).nonzero()[0]))
```
