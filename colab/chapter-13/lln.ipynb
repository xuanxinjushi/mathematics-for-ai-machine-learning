# Law of Large Numbers Simulator

Chapter 13: Probability and Random Variables · *Mathematics for AI and Machine Learning* by Fuheng Wu

Track empirical means converging to expectation as sample size grows.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-13/lln/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# The demo's defaults: Bernoulli(p) coin, n flips, seed 42
# (same parameters; torch's random draws differ from the page's generator)
p, n, seed = 0.3, 1200, 42
torch.manual_seed(seed)

flips = (torch.rand(n, dtype=torch.float64) < p).double()
running_mean = flips.cumsum(0) / torch.arange(1, n + 1, dtype=torch.float64)

# ── The sample mean approaches E[X] = p ─────────────────────────────────────
print("   k    mean of first k    |error|     sqrt(p(1-p)/k)")
for k in (10, 50, 100, 300, 600, 1200):
    m = running_mean[k - 1].item()
    print(f"{k:5d}   {m:.4f}             {abs(m - p):.4f}      {(p * (1 - p) / k) ** 0.5:.4f}")

# ── The typical error shrinks like 1/sqrt(n): check over many runs ──────────
runs = 5000
print("\n    n    average |mean - p| over", runs, "runs    x sqrt(n)")
for m in (100, 400, 1600):
    means = (torch.rand(runs, m, dtype=torch.float64) < p).double().mean(1)
    err = (means - p).abs().mean().item()
    print(f"{m:5d}    {err:.5f}                          {err * m ** 0.5:.4f}")
print("(constant ~ sqrt(2/pi) * sqrt(p(1-p)) =", ((2 / torch.pi) * p * (1 - p)) ** 0.5, ")")
```
