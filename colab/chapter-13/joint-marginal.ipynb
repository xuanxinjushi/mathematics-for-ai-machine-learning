# Joint vs Marginal PMF

Chapter 13: Probability and Random Variables · *Mathematics for AI and Machine Learning* by Fuheng Wu

Heatmap a discrete joint table and inspect row/column marginals.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-13/joint-marginal/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.set_printoptions(precision=4)

# The demo's defaults: four free cells, the other five share the rest equally
a, b, c, d = 0.25, 0.15, 0.10, 0.20
share = max(0.05, 1 - (a + b + c + d)) / 5
P = torch.tensor([[a, b, share],
                  [c, d, share],
                  [share, share, share]], dtype=torch.float64)
P = P / P.sum()                         # rows are X = 0, 1, 2; columns are Y = 0, 1, 2
print("joint P(X, Y) =\n", P)
print("sum =", P.sum().item())

# ── Marginals: sum out the other variable ───────────────────────────────────
pX = P.sum(dim=1)
pY = P.sum(dim=0)
print("\nP(X) =", pX.tolist())
print("P(Y) =", pY.tolist())

# ── Conditionals: one row (or column) renormalized ──────────────────────────
print("P(Y | X = 0) =", (P[0] / pX[0]).tolist())
print("P(X | Y = 1) =", (P[:, 1] / pY[1]).tolist())

# Different joints can share the same marginals: the product P(X)P(Y) is one
print("\nP(X)P(Y) has the same marginals:",
      torch.allclose(torch.outer(pX, pY).sum(1), pX), torch.allclose(torch.outer(pX, pY).sum(0), pY))
print("but equals the joint?", torch.allclose(torch.outer(pX, pY), P))

# ── Expectations from the table ─────────────────────────────────────────────
vals = torch.arange(3, dtype=torch.float64)
EX, EY = (vals * pX).sum(), (vals * pY).sum()
EXY = (torch.outer(vals, vals) * P).sum()
print(f"\nE[X] = {EX.item():.4f}, E[Y] = {EY.item():.4f}, Cov(X, Y) = {(EXY - EX * EY).item():.4f}")
```
