# Bayes Rule Diagnostic Test

Chapter 13: Probability and Random Variables · *Mathematics for AI and Machine Learning* by Fuheng Wu

See how base rate, sensitivity, and specificity determine posterior probability.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-13/bayes-test/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

torch.manual_seed(0)

# The demo's defaults: prevalence 5%, sensitivity 95%, specificity 90%
prev, sens, spec = 0.05, 0.95, 0.90

# ── Bayes' rule: P(D | +) = P(+ | D) P(D) / P(+) ────────────────────────────
p_pos = sens * prev + (1 - spec) * (1 - prev)
post = sens * prev / p_pos
print(f"P(+) = {p_pos:.4f}")
print(f"P(disease | positive test) = {post:.4f}")
print(f"P(no disease | negative)   = {spec * (1 - prev) / (1 - p_pos):.4f}")

# Natural frequencies for 10,000 people
n = 10000
tp, fn = sens * prev * n, (1 - sens) * prev * n
fp, tn = (1 - spec) * (1 - prev) * n, spec * (1 - prev) * n
print(f"\nof {n}: {tp:.0f} true positives, {fp:.0f} false positives -> {tp:.0f}/{tp + fp:.0f} = {tp / (tp + fp):.4f}")

# ── Simulate a million people to check ──────────────────────────────────────
N = 1_000_000
sick = torch.rand(N) < prev
positive = torch.where(sick, torch.rand(N) < sens, torch.rand(N) < 1 - spec)
print("simulated P(disease | positive) =", (sick & positive).sum().item() / positive.sum().item())

# ── The base rate dominates ─────────────────────────────────────────────────
print("\n prevalence   P(disease | +)")
for p in (0.001, 0.01, 0.05, 0.2, 0.5):
    print(f"   {p:6.3f}       {sens * p / (sens * p + (1 - spec) * (1 - p)):.4f}")
# A second independent positive test: use the posterior as the new prior
post2 = sens * post / (sens * post + (1 - spec) * (1 - post))
print("\nafter two positive tests:", round(post2, 4))
```
