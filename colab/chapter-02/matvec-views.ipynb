Compare row-dot-product and column-linear-combination interpretations.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-2/matvec-views/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

A = torch.tensor([[1.2, -0.7], [0.9, 1.7]])
x = torch.tensor([1.5, -0.8])

# Matrix–vector product y = A x
y = A @ x

print("A @ x =", y.tolist())

# Row view: y_i = (row i of A) · x
print("row view:", (A[0] @ x).item(), (A[1] @ x).item())

# Column view: y = x_1 (col 1) + x_2 (col 2)  — linear combination of columns
print("col view:", (x[0] * A[:, 0] + x[1] * A[:, 1]).tolist())
```
