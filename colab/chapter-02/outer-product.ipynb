Construct a rank-1 matrix from two vectors and inspect structure.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-2/outer-product/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

u = torch.tensor([1.4, -0.9])
v = torch.tensor([2.0, 0.6])

# Rank-1 matrix: A[i,j] = u_i v_j  (only one independent row/column pair)
A = torch.outer(u, v)

print(A)
print("trace:", A.trace().item())           # sum of diagonal entries
print("det:", torch.linalg.det(A).item())   # 0 for a 2×2 rank-1 matrix
```
