```python
import torch

A = torch.tensor([[1.1, 0.7], [-0.8, 1.9]])
B = torch.tensor([[1.9, -0.8], [0.7, 1.1]])

# Matrix product C = A B
# Entry c_ij = (row i of A) · (column j of B) = Σ_k a_ik b_kj
C = A @ B

print("A @ B =\n", C)
print("c_11 =", (A[0] @ B[:, 0]).item())
```
