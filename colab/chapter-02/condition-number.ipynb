# Condition Number and Sensitivity

Chapter 2: Matrix · *Mathematics for AI and Machine Learning* by Fuheng Wu

See how near-singular matrices amplify perturbations in linear solves.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-2/condition-number/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

# Nearly dependent rows → A is ill-conditioned (large κ₂)
eps = 0.15
A = torch.tensor([[1.0, 1.0], [1.0, 1.0 + eps]])
b = torch.tensor([0.0, 0.0])

# Solve A x = b; tiny changes in A or b can swing x when κ₂ is large
x = torch.linalg.solve(A, b)

# Condition number κ₂(A) = ||A||₂ ||A⁻¹||₂  (amplifies input/output errors)
kappa = torch.linalg.cond(A)

print("kappa_2(A) =", kappa.item())
print("x =", x.tolist())
```
